```python

from flask import Flask, render_template
import logging
import platform
import sys

# Initialize the Flask application
app = Flask(__name__)

# Configure application logging
logging.basicConfig(
    level=logging.INFO,
    format="%(levelname)s: %(message)s"
)

logger = logging.getLogger("hello_world_app")


# Define the homepage
@app.route("/")
def home():
    logger.info("Homepage requested")
    return render_template("index.html")


# Define an application information page
@app.route("/about")
def about():
    return {
        "application": "Hello World",
        "framework": "Flask",
        "python_version": platform.python_version(),
        "environment": "Jupyter Notebook" if is_jupyter() else "Python"
    }


# Detect the execution environment
def is_jupyter():
    try:
        from IPython import get_ipython

        shell = get_ipython()

        return (
            shell is not None
            and shell.__class__.__name__ == "ZMQInteractiveShell"
        )
    except (ImportError, NameError):
        return False


# Start the development server
if __name__ == "__main__":
    logger.info("Starting Hello World application")

    if is_jupyter():
        app.run(
            host="127.0.0.1",
            port=5000,
            debug=False,
            use_reloader=False
        )
    else:
        app.run(
            host="127.0.0.1",
            port=5000,
            debug=True,
            use_reloader=False
        )
```
